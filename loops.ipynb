```python
import random

low_num = 1
high_num = 100

correct_num = random.randint(low_num, high_num)

insert_num = int(input("Enter your number: "))

while insert_num != correct_num:

    if insert_num > correct_num:
        print("The number is too high")

    elif insert_num < correct_num:
        print("The number is too low")

    insert_num = int(input("Enter your number: "))

print("Congratulations!! The correct number is:", correct_num)
```
